یک برنامه پایتون بنویسید که از کاربر نام، سن و کشور مبدا را بپرسد و سپس:

```python
username = input("please inter your name: ")
age= int(input("please inter age: "))
country = input("please inter your country: ")
```

-	بررسی کنید که آیا کاربر بزرگسال است (18 سال یا بیشتر).

```python
customer = age>=18
```

-	بررسی کنید که آیا کاربر اهل ایالات متحده است یا خیر.

```python
where = country.lower()=="Iran"
```

-	بررسی کنید که آیا نام کاربر با یک مصوت (A، E، I، O، U) شروع می شود.

```python
start= username[0].lower() in {"a","e","i","o","u"}
```

-	بر اساس این بررسی ها، برنامه باید برای هر ترکیب ممکن از شرایط، پیام متفاوتی را چاپ کند.

```python
if customer and where and start:
     print("welcome, you are accept")
elif customer and where:
     print ("accept first and second of sentences")
#elif customer and start:
     print ("accept first and thrid of sentences")
elif where:
     print("the country is accept")
elif customer:
     print("accept name")
else:
     print("don't accept")
```
